```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/sample_-_superstore.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python
df.drop_duplicates(inplace=True)
print("Shape:", df.shape)
```

```python
high_sales = df[df['Sales'] > 500]
high_sales.head()
```

```python
technology = df[df['Category'] == 'Technology']
technology.head()
```

```python
sorted_sales = df.sort_values(by='Sales', ascending=False)
print(sorted_sales[['Sales', 'Profit']].head())
```

```python
print("Total Sales:", df['Sales'].sum())
```

```python
print("Total Profit:", df['Profit'].sum())
```

```python
print("Average Sales:", df['Sales'].mean())
```

```python
print("Average Profit:", df['Profit'].mean())
```

```python
print("Maximum Sales:", df['Sales'].max())
```

```python
print("Minimum Sales:", df['Sales'].min())
```

```python
region_sales = df.groupby('Region')[['Sales','Profit']].sum()
print(region_sales)
```

```python
category_sales = df.groupby('Category')[['Sales','Profit']].sum()
print(category_sales)
```

```python
category_sales['Sales'].plot(kind='bar', figsize=(6,4))
plt.title("Category-wise Sales")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
monthly_sales = df.groupby('Order Date')['Sales'].sum()

monthly_sales.plot(figsize=(10,5))
plt.title("Sales Trend")
plt.xlabel("Order Date")
plt.ylabel("Sales")
plt.show()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
monthly_sales = df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum()

monthly_sales.plot(figsize=(10,5))
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
plt.hist(df['Sales'], bins=20)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.scatter(df['Sales'], df['Profit'])
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```
