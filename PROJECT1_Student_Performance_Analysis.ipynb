```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/Student_performance_data _.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
print(df.isnull().sum())
```

```python
for col in df.columns:
    if df[col].dtype == "object":
        df[col].fillna(df[col].mode()[0])
    else:
        df[col].fillna(df[col].mean())
```

```python
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
print(df.shape)
```

```python
sorted_df = df.sort_values(by='GPA', ascending=False)
print("Top 5 Students with Highest Exam Scores:")
sorted_df.head()
```

```python
print(df.columns)
```

```python
high_score = df[df['GPA'] > 3.0]
print("\nStudents with GPA > 3.0")
high_score.head()
```

```python
print(df.groupby('Gender')['GPA'].mean())
```

```python
print(df.groupby('ParentalEducation')['GPA'].mean())
```

```python
plt.figure(figsize=(6,4))
plt.hist(df['GPA'], bins=10)
plt.title("GPA Distribution")
plt.xlabel("GPA")
plt.ylabel("Frequency")
plt.show()
```

```python
gender_avg = df.groupby('Gender')['GPA'].mean()
plt.figure(figsize=(5,4))
plt.bar(gender_avg.index, gender_avg.values)
plt.title("Average GPA by Gender")
plt.xlabel("Gender")
plt.ylabel("Average GPA")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df['StudyTimeWeekly'], df['GPA'])
plt.xlabel("Study Time Weekly")
plt.ylabel("GPA")
plt.title("Study Time Weekly vs GPA")
plt.show()
```

```python
avg = df.groupby('Gender')['GPA'].mean()
plt.figure(figsize=(5,4))
plt.plot(avg.index, avg.values, marker='o')
plt.title("Average GPA by Gender")
plt.ylabel("Average GPA")
plt.show()
```
