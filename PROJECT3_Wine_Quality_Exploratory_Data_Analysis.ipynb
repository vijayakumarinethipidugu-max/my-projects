```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df_red = pd.read_csv("/content/winequality-red.csv", delimiter=';')
df_white = pd.read_csv("/content/winequality-white.csv", delimiter=';')

df_red['source'] = 'red'
df_white['source'] = 'white'

df = pd.concat([df_red, df_white], ignore_index=True)
```

```python
numeric = df.select_dtypes(include=np.number)
```

```python
print("\nMean")
print(numeric.mean())
```

```python
print("\nMedian")
print(numeric.median())
```

```python
print("\nMode")
print(numeric.mode().iloc[0])
```

```python
print("\nVariance")
print(numeric.var())
```

```python
print("\nStandard Deviation")
print(numeric.std())
```

```python
print("\nCorrelation Matrix")
corr = numeric.corr()
corr
```

```python
print("\nHighly Correlated Features (>|0.7|)")
high_corr = corr.unstack()
high_corr = high_corr[(abs(high_corr) > 0.7) & (abs(high_corr) < 1)]
print(high_corr)
```

```python
for col in numeric.columns:
    Q1 = numeric[col].quantile(0.25)
    Q3 = numeric[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = numeric[(numeric[col] < lower) | (numeric[col] > upper)]

    print(col, ":", len(outliers))
```

```python
plt.figure(figsize=(6,4))
plt.hist(df["quality"], bins=10)
plt.title("Quality Distribution")
plt.xlabel("Quality")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df["alcohol"], df["quality"])
plt.title("Alcohol vs Quality")
plt.xlabel("Alcohol")
plt.ylabel("Quality")
plt.show()
```

```python
plt.figure(figsize=(6,4))
df["quality"].value_counts().sort_index().plot(kind="line")
plt.title("Wine Quality Count")
plt.xlabel("Quality")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6,4))
df["quality"].value_counts().sort_index().plot(kind="bar")
plt.title("Quality Frequency")
plt.xlabel("Quality")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6,6))
df["quality"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Wine Quality Distribution")
plt.ylabel("")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.hist(df["alcohol"], bins=10, color="orange", edgecolor="black")
plt.title("Alcohol Distribution")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df["alcohol"], df["quality"], color="green")
plt.title("Alcohol vs Quality")
plt.xlabel("Alcohol")
plt.ylabel("Quality")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.boxplot(df["alcohol"])
plt.title("Alcohol Box Plot")
plt.ylabel("Alcohol")
plt.show()
```
